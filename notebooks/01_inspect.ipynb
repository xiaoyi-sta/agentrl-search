```python
import pandas as pd
from pathlib import Path
import os

# 定位项目根（防止 cwd 在 notebooks/ 里）
ROOT = Path.cwd()
while not (ROOT / "data").exists() and ROOT != ROOT.parent:
    ROOT = ROOT.parent
os.chdir(ROOT)

df = pd.read_parquet("data/raw/hotpot_dev.parquet")

def to_paras(ctx):
    if hasattr(ctx, "to_pylist"):
        ctx = ctx.to_pylist()
    if isinstance(ctx, dict):
        return [{"title": t, "sentences": list(s)} for t, s in ctx.items()]
    out = []
    for item in ctx:
        if isinstance(item, dict):
            out.append({"title": item["title"], "sentences": item["sentences"]})
        else:
            out.append({"title": item[0], "sentences": item[1]})
    return out

def to_support(sf):
    if hasattr(sf, "to_pylist"):
        sf = sf.to_pylist()
    if isinstance(sf, dict):
        return [{"title": t, "sent_id": i} for t, i in zip(sf["title"], sf["sent_id"])]
    return [{"title": x["title"], "sent_id": x["sent_id"]} for x in sf]

r = df.iloc[0]
print(to_paras(r["context"])[0]["title"], len(to_paras(r["context"])))
print(to_support(r["supporting_facts"]))
```
