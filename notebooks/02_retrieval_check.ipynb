跑完 `scripts/build_data.py` 后，这里看语料统计和检索效果。

```python
import json, sys
from pathlib import Path

ROOT = Path.cwd()
while not (ROOT / 'data').exists() and ROOT != ROOT.parent:
    ROOT = ROOT.parent
sys.path.insert(0, str(ROOT / 'src'))

stats = json.loads((ROOT / 'data/processed/stats.json').read_text())
print(json.dumps(stats, ensure_ascii=False, indent=2))
```

```python
from agentrl.retrieval.bm25 import BM25Index

idx = BM25Index.from_jsonl(ROOT / 'data/processed/corpus.jsonl')
q = "Were Scott Derrickson and Ed Wood of the same nationality?"
for d in idx.search(q, k=3):
    print(round(d['score'], 2), d['title'], '|', d['text'][:120])
```

```python
from agentrl.retrieval.bm25 import evaluate
import json as _json
qs = [_json.loads(l) for l in open(ROOT/'data/processed/splits/eval200.jsonl')]
print(evaluate(idx, qs[:100], k=3)['recall_at_k'])
```
